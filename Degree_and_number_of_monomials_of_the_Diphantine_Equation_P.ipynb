```python
from sympy import Poly
from sympy import Mul
from collections import Counter

def count_monomials_by_degree(expr, degree_range=range(1, 50)):
    poly = Poly(expr)
    monomials = poly.terms()
    # Compute total degree of each monomial and track the one with the highest degree
    highest_deg = 0
    highest_deg_monomial = None

    for monomial, coeff in monomials:
        total_deg = sum(monomial)
        if total_deg > highest_deg:
            highest_deg = total_deg
            highest_deg_monomial = (monomial, coeff)

    degree_counter = Counter(sum(m) for m, _ in monomials)
    monomial_expr = highest_deg_monomial[1] * Mul(*[gen**exp for gen, exp in zip(poly.gens, highest_deg_monomial[0]) if exp > 0])

   #print(highest_deg_monomial)
    total = []
    solution = {deg: degree_counter.get(deg, 0) for deg in degree_range}
    for values in solution.items():
      total.append(values[1])
    return([solution, f"Total monomials: {sum(total)}", f"Highest degree monomial: {monomial_expr}"])
```
